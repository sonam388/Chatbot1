```python
# Naive Bayes ---> SML ---> probability 

# Multinomial NB   -> text ---> 

# Email Spam Detaction System ----> 

# win money now ----> spam
# hellow ----> normal
# project discussion today -----> Normal
# claim your free prize ----> Smap

# NLP --> Part 

from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import TfidfVectorizer

# TfidfVectorizer ---> is used to convert text data into numbaric data

# number 

# email/text ----> number 


from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import TfidfVectorizer

email = [
    "Win money now",
    "Free Lottery Ticket",
    "metting schudle tomorrow",
    "Project Discission today",
    "clam your free prize"
]
label = ["Smap","Spam","Normal","Normal","Spam"]
vectorizer = TfidfVectorizer()
x = vectorizer.fit_transform(email)
model = MultinomialNB()
model.fit(x,label)
text = ["you lottery is here"]
x_predict = vectorizer.transform(text)
print(model.predict(x_predict))
```
