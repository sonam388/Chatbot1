```python
# Filtering ---> 

import pandas as pd

data = {
    "Product":["1+" , "Vivo" , "MI" , "HP" , "DEll" , "1+"],
    "Price":[12000,22000,13000,44000,16000,10000]
}
df = pd.DataFrame(data)
# price > 20000
con = df['Price'] < 20000
con2 = df['Price'] > 15000

df = df[con & con2]

print(df)
```

```python
# iloc ---> index location

import pandas as pd

data = {
    "Product":["1+" , "Vivo" , "MI" , "HP" , "DEll" , "1+"],
    "Price":[12000,22000,13000,44000,16000,10000]
}
df = pd.DataFrame(data)

# index --> row --->  columns --->  value

print(df.iloc[0])  # Row
```
