```python
# Filtering ---> 
# We can filter data by using 
# Conitions (Relational Operator )
# loc[] , iloc[] 

import pandas as pd

d = {
    "Product":["1+","Vivo", "HP","HP"],
    "Price":[20000,30000,10000,14000]
}

df = pd.DataFrame(d)

con = df['Price'] <= 20000 
con2 = df['Product'] == "HP"

df2 = df[con & con2]

print(df2)
```
