```python
import json 
import random

from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import TfidfVectorizer
import joblib

with open("data.json" , "r") as file:
    data = json.load(file)

# print(data)

# Arange Data

# patten = [hii]
# tag = [g]
# res = {tag : responce}

sentance = [] # patterns
lables = []  # tag
responce =  {}  # tag : responce

for i in data['intents']:
    tag = i['tag']
    responce[tag] = i["responses"]
    for patten in i['patterns']:
        sentance.append(patten)
        lables.append(tag)

# print(sentance)
# print(lables)
# print(responce)


vector = TfidfVectorizer()

x = vector.fit_transform(sentance)

model = MultinomialNB()

model.fit(x,lables)

joblib.dump(model ,"Chatbot.pkl")
joblib.dump(vector , "vector.pkl")
joblib.dump(responce , "res.pkl")
```

```python
import joblib
import random

model = joblib.load("Chatbot.pkl")
vector = joblib.load("vector.pkl")
res = joblib.load("res.pkl")

while True:
    user = input("You : ")
    if user == 'exit':
        break
    user_vector = vector.transform([user]).toarray()
    predict = model.predict(user_vector)[0]  # ['python'] --> python
    if predict in res:
        reply = random.choice(res[predict])
        print("Bot : ",reply)
```

```python
# TFIDF Vectorizer ----> 

from sklearn.feature_extraction.text import TfidfVectorizer
sentance = [
    "I like Python",
    "I like Java",
    "Python is easy"
]
vector = TfidfVectorizer()

x = vector.fit_transform(sentance)

print(vector.get_feature_names_out())
print(x.toarray())
```

```python
# CountVectorizer -->  NLP

from sklearn.feature_extraction.text import CountVectorizer

data = [
    "i like python",
    "i like python",
    "python is simple"
]
vector = CountVectorizer()

x = vector.fit_transform(data)

print(vector.get_feature_names_out())
print(x.toarray())
```
