```python
import pandas as pd

# sorting --> 

# sort_values() --> 

d = {
    "Name":["Kunal","Dev","Aman"],
    "age":[22,11,33]
}

i = [22,11,33]

df = pd.DataFrame(d , index=i)

# df2 = df.sort_values("age",ascending=1)

df2 = df.sort_index()

# sort_index() ---> 

print(df2)
```

```python
# concat(p) --> 

a = [[1,2],[3,4]]
b = [[5,6],[7,8]]

c = ["A","B"]
c1= ["X","Y"]

df1 = pd.DataFrame(a , columns=c)
df2 = pd.DataFrame(b , columns=c1)

result = [df1 , df2]

df3 = pd.concat(result , axis=1)

print(df3)
```

```python
# Adding & Droping Columns 

d = {
    "Name":["Kunal","Dev","Aman"],
    "age":[22,11,33]
}


# con

# dropna()

df = pd.DataFrame(d)

df['Subject'] = "Python"

df2 = df.drop(columns=["Name","age"])

print(df2)
```
