```python
# PCA --> 

import pandas as pd
# from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LinearRegression

data = { "StudyHours": [8, 6, 9, 3, 4], 
         "Attendance": [90, 80, 95, 60, 65], 
         "Assignment": [85, 75, 90, 55, 60], 
         "TestScore": [88, 78, 92, 58, 62] } 


df = pd.DataFrame(data) 
x = df[["StudyHours","Attendance","Assignment"]]
y = df[['TestScore']]

# s = StandardScaler()

# x_scaler = s.fit_transform(x)

pca = PCA(n_components=1)

x_pca = pca.fit_transform(x)
# print(x_pca)
model = LinearRegression()
model.fit(x_pca , y)

new_value = [[7,60,50]]

new_value_pca = pca.transform(new_value)

print(model.predict(new_value_pca))
```
