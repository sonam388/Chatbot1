```python
d1 = {
    "Name":["Kunal","Aman","Dev"],
    "id":[1,2,3]
}

d2 = {
    "Subject":["Python","HTML","CSS"],
    "id":[1,2,4]
}

df1 = pd.DataFrame(d1)
df2 = pd.DataFrame(d2)

result = pd.merge(df1, df2 , on="id", how="outer")

print(result)


# inner return Only Matching records

# left returns all records from the left DataFrame and return match recors from right 

# right returns all records from the right DataFrame and return match recors from left 

# outer return all record from both dataframes
```
