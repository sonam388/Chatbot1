```python
# Smart trafic Signal   ---> 

# law    10
# medium  20
# high    30

import random

traffic = ["Low","Medium","High"]

actions = [10,20,30]

# Q-table

Q = {}


for state in traffic:
    Q[state] = {}

    for action in actions:
        Q[state][action] = 0

# print(Q)


# train CNN RNN


for episode in range(1000):
    state = random.choice(traffic)
    action = random.choice(actions)

    # Rewards 

    if state == "Low":
        if action ==10:
            rewards = 10
        else:
            rewards = -2
            
    elif state == "Medium":
        if action == 20:
            rewards = 10
        else:
            rewards = -5
    else :
        if action == 30:
            rewards = 10
        else:
            rewards = -10

    Q[state][action] = Q[state][action] + 0.1 * (
        rewards - Q[state][action]
    )


# "Low",[10] 

# new Q = old Q + 0.1 * (r - old Q)
# new Q = 0 + 0.1 * (10 - 0)  --> 1

# "Low"[10]
# new Q = old Q + 0.1 * (r - old Q)
#       = 1 + 0.1 * (9)    1.9






print("Trainning Complete")

print(Q)


for state in traffic:
    best_action = max(Q[state],key=Q[state].get)
    print(
    state,
    "Traffic ---> Green for ",
    best_action,
    "Seconds"
)
    
```
